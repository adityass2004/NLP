# Import Library

```python
from transformers import pipeline
```

# Load Summarization Model

```python
summarizer = pipeline(
    task="text-generation",
    model="google/flan-t5-base",
    model_kwargs={"tie_word_embeddings": False}
)
```

# Test It


```python
text = """
Artificial Intelligence is transforming industries worldwide.
It helps automate tasks, improve decisions,
and increase efficiency in business operations.
"""

prompt = "Summarize this text: " + text

result = summarizer(
    prompt,
    max_length=50
)

print(result[0]["generated_text"])
```

# Final


```python
while True:
    text = input("Enter text (type exit to stop): ")

    if text.lower() == "exit":
        break

    prompt = "Summarize this text: " + text

    result = summarizer(
        prompt,
        max_length=60
    )

    print("\nSummary:", result[0]["generated_text"])
```
