# Import Libraries

```python
from transformers import pipeline
```

# Load ChatBot Model

```python
chatbot = pipeline(
    "text-generation",
    model="gpt2"
)
```

# Give Input to Chatbot

```python
user_input = "Hello, how are you?"
```

# Generate Response

- Single chat  

```python
response = chatbot(
    user_input,
    max_length=50,
    num_return_sequences=1
)

print(response[0]["generated_text"])
```

- Loop Chat

```python
name = input("Enter your name: ")

conversation_history = []

print(f"AI Bot: Hello {name}! Type 'exit' to stop chatting.")

while True:
    user_input = input(f"{name}: ")

    if user_input.lower() == "exit":
        print("AI Bot: Goodbye!")
        break

    conversation_history.append(f"{name}: {user_input}")

    context = "\n".join(conversation_history[-5:])

    response = chatbot(
        context,
        max_length=len(context.split()) + 40,
        temperature=0.8,
        do_sample=True,
        top_k=50,
        top_p=0.95,
        num_return_sequences=1
    )

    generated_text = response[0]["generated_text"]

    bot_reply = generated_text[len(context):].strip()

    print("AI Bot:", bot_reply)

    conversation_history.append(f"AI Bot: {bot_reply}")
```
